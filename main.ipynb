# Functions

```python
import random

skills = {
    "1": {"name": "attack", "type": "damage", "value": (10, 20)},
    "2": {"name": "punchofluck", "type": "damage", "value": (0, 50)},
    "3": {"name": "heal", "type": "heal", "value": 25},
    "4": {"name": "defend", "type": "defend"}
}

player_hp = 100
enemy_hp = 100
player_status = {"defend": False}
```

```python
def get_player_action():
    while True:
        print("\nChoose action:")
        print("[1] Attack ⚔️")
        print("[2] Punch of Luck 🍀")
        print("[3] Heal 💊")
        print("[4] Defend 🛡️")

        choice = input("Enter number (1-4): ").strip()

        if choice in skills:
            return choice
        else:
            print("❌ Invalid input! Try again.")
```

# Game Logic

```python
def main():
    player_hp = 100
    enemy_hp = 100
    player_status = {"defend": False}

    turn = 1

    print("⚔️ === Turn-Based Battle Game === ⚔️")

    while player_hp > 0 and enemy_hp > 0:
        print(f"\n--- Turn {turn} ---")
        print(f"❤️ Player HP: {player_hp} | 👾 Enemy HP: {enemy_hp}")

        # ===== Player Turn =====
        choice = get_player_action()
        skill = skills[choice]

        if skill["type"] == "damage":
            damage = random.randint(*skill["value"])
            enemy_hp -= damage
            print(f"\n⚔️ You used {skill['name']} and deal {damage} damage!")

        elif skill["type"] == "heal":
            player_hp = min(player_hp + skill["value"], 100)
            print(f"\n💊 You heal for {skill['value']} HP!")

        elif skill["type"] == "defend":
            player_status["defend"] = True
            print("\n🛡️ You are defending this turn!")

        # ===== Check Enemy Dead =====
        if enemy_hp <= 0:
            print("\n🎉 You Win!")
            break

        # ===== Enemy Turn =====
        if enemy_hp < 30:
            enemy_action = "heal"
        else:
            enemy_action = random.choice(["attack", "attack", "heal"])

        if enemy_action == "attack":
            damage = random.randint(10, 30)

            if player_status["defend"]:
                damage //= 2
                player_status["defend"] = False
                print("🛡️ You reduced the damage!")

            player_hp -= damage
            print(f"👾 Enemy attacks and deals {damage} damage!")

        else:
            heal = 15
            enemy_hp = min(enemy_hp + heal, 100)
            print(f"👾 Enemy heals for {heal} HP!")

        # ===== Check Player Dead =====
        if player_hp <= 0:
            print("\n💀 You Lose!")
            break

        turn += 1

    print("\n=== END ===")
```

# Run Program

```python
main()
```
